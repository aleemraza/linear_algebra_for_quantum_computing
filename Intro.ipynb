Span
The span of vectors is the set of all possible vectors you can create using linear combinations.

```python
import numpy as np

u = np.array([1, 0])
v = np.array([0, 1])

# Create matrix using vectors as columns
A = np.column_stack((u,v))
# Calculate rank
rank = np.linalg.matrix_rank(A)
print("Matrix:")
print(A)
print("Rank:", rank)
```

Basis
A basis is a set of vectors that can generate the entire space through linear combinations, while being linearly independent.

```python
import numpy as np

u = np.array([1, 0])
v = np.array([0, 1])
w = np.array([5, 3])

# Matrix containing basis vectors
A = np.column_stack((u, v))
# Solve A @ x = w
coefficients = np.linalg.solve(A, w)
print(coefficients)
# Verify
result = A @ coefficients

print("Reconstructed vector:", result)
```

```python
import numpy as np
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)
alpha = 1 / np.sqrt(2)
beta = 1 / np.sqrt(2)
psi = alpha * zero + beta * one
normalization = np.vdot(psi, psi)
# -----------------------------------
# Step 5: Calculate probabilities
# -----------------------------------

p_zero = abs(alpha) ** 2
p_one = abs(beta) ** 2


# -----------------------------------
# Step 6: Display results
# -----------------------------------

print("Basis state |0> =", zero)
print("Basis state |1> =", one)

print("\nAlpha =", alpha)
print("Beta  =", beta)

print("\nQuantum state |psi> =", psi)

print("\nNormalization =", normalization)

print("\nProbability of measuring 0 =", p_zero)
print("Probability of measuring 1 =", p_one)

print("\nTotal probability =", p_zero + p_one)
```

```python
import numpy as np

# Basis states
zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)

# Amplitudes
alpha = np.sqrt(3 / 4)
beta = np.sqrt(2 / 4)

# Quantum state
psi = alpha * zero + beta * one

# Normalization
normalization = np.vdot(psi, psi)

# Probabilities
p_zero = abs(alpha) ** 2
p_one = abs(beta) ** 2

# Results
print("Alpha =", alpha)
print("Beta =", beta)

print("\nQuantum state:")
print(psi)

print("\nNormalization =", normalization)

print("\nP(0) =", p_zero)
print("P(1) =", p_one)

print("\nTotal probability =", p_zero + p_one)
```
